## 🐍 Manual Gradient Descent for Simple Linear Regression

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

### `Step 1: Load dataset`

```python
df = pd.read_csv("Study Hour.csv")
df.sample(5)
```

```python
X = df["Hours"].values
y = df["Scores"].values
```

### `Step 2: Initialize parameters`

```python
n = len(X) # Reshape for easy computation - 25 rows
m = 0     # slope
b = 0     # intercept
alpha = 0.01   # learning rate
epochs = 1000  # number of iterations
```

### `Step 3: Gradient Descent loop`

```python
for _ in range(epochs):
    y_pred = m * X + b
    error = y - y_pred
    
    # Compute gradients
    dm = (-2/n) * np.sum(X * error)
    db = (-2/n) * np.sum(error)
    
    # Update parameters
    m = m - alpha * dm
    b = b - alpha * db
```

```python
print(f"Final slope (m): {m}")
print(f"Final intercept (b): {b}")
```

```python
df['Predicted_Y'] = m * X + b
print(df.head())
```

```python
# Step 5: Plot Actual vs Predicted with Best Fit Line
plt.scatter(X, y, color='blue', label='Actual Scores')         # scatter plot
plt.plot(X, df['Predicted_Y'], color='red', label='Best Fit Line')  # regression line
plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.title("Linear Regression (Manual Gradient Descent)")
plt.legend()
plt.show()
```
