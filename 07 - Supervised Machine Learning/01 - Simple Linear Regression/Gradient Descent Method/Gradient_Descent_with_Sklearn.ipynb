## 🐍 2. Gradient Descent Using scikit‑learn

#### `sklearn.LinearRegression uses the closed‑form OLS solution by default (not gradient descent)`
##### `But sklearn has other models (like SGDRegressor) that use Stochastic Gradient Descent.`

```python
import pandas as pd
from sklearn.linear_model import SGDRegressor
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("Study Hour.csv")
df.sample(5)
```

```python
X = df[['Hours']]
y = df['Scores']
```

### `Step 2: Create model with gradient descent`

```python
model = SGDRegressor(max_iter=1000, eta0=0.01, learning_rate = 'constant')
model.fit(X,y)
```

### `Step 3: Get slope (m) and intercept (b)`

```python
m = model.coef_[0]
b = model.intercept_[0]
```

```python
print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
```

```python
y_pred = model.predict(X)
df['Predicted_Y'] = y_pred
print(df.head())
```

```python
plt.scatter(X,y, color='blue', label='Actual Scores')
plt.plot(X, y_pred, color='red', label='Best Fit Line based on - Predicted Score')
plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.title("Linear Regression (SGD Gradient Descent By SKLEARN)")
plt.legend()
plt.show()
```
