## `Linear Regression Scratch By a real-word example.`

Let’s say you are a bakery owner. You want to predict how much money you will make in a day based on the number of cupcakes you sell.

To predict this, you will gather data on the number of cupcakes you sell and how much money you earned in several days. Then, you will use a Linear Regression algorithm to find the best-fit line that describes the relationship between the number of cupcakes sold and the amount of money made.

Once you have the best-fit line, you can use it to predict how much money you will make for a different number of cupcakes sold. For example, if the line tells you that you make $50 by selling 20 cupcakes, then you can predict that selling 40 cupcakes will make $100 in a day.

That’s how the Linear Regression algorithm works. It finds the best-fit line that describes the relationship between the input variable and the output variable so we can predict the output variable based on a new input variable.

- Let’s first import the necessary Python libraries and create a sample data based on the example we discussed above:

### `Import Libraries`

- Import `Linear Regression` - From `linear_model` under `sklearn`

```python
import numpy as np
import pandas as pd
import plotly.graph_objs as go
from sklearn.linear_model import LinearRegression
```

### `Create Sample Data` 

```python
num_cupcakes = [10, 20, 30, 40, 50]
money_made = [25, 50, 75, 100, 125]
```

```python
# No need - this CODE PART - just make those data into a dataframe. 
data = pd.DataFrame({
    'num_cupcakes': num_cupcakes,
    'money_made': money_made
})
data
```

```python
# Not Needed - if you want to just transfer the dataframe into csv - then do it.
data.to_csv('cup_data.csv')
```

```python
figure = go.Figure()
figure.add_trace(go.Scatter(x = num_cupcakes, y = money_made, mode='markers'))
figure.update_layout(title= "Relationship Between CupCakes Sales and Money Made", 
                    xaxis_title='Number of Cupcakes Sold (Independent Variable)',
                    yaxis_title='Money Made (Dependent Variable)')
figure.show()
```

### `create a linear regression model`

```python
x = np.array(num_cupcakes).reshape((5,1))
y = np.array(money_made)
model = LinearRegression().fit(x,y)
```

- The reshape((-1, 1)) function is used to prepare your data. In simple terms, it takes a list of numbers (a 1-dimensional array) and converts it into a 2-dimensional array with a single column. Technically, the 1 specifies one column, and the -1 is a placeholder that tells NumPy to automatically calculate the necessary number of rows to accommodate all the data.
- The -1 means "figure out how many rows are needed," and the 1 means "make sure there's only one column."

- ``How we can predict the value of a dependent variable using the value of an independent variable using our Machine Learning model``

### `predict the money made for UNSEEN cupcakes sold`

```python
new_num_cupcakes = [[200]]   #Unseen Unknown data 
new_money_made = model.predict(new_num_cupcakes)
print("Predicted money made for cupcakes sold: ", new_money_made)
```

- The double square brackets [[78]] are used because the model.predict() method, like the fit() method, expects its input to be a 2-dimensional array (a matrix), even when you're predicting for just a single value. The outer [] signifies the array itself, while the inner [78] represents a single row (or sample) containing the value 78, thus maintaining the expected 2D structure of features for the prediction.
