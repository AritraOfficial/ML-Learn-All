### 🐍 OLS Method - Simple Linear Regression with sklearn

#### `sklearn.LinearRegression uses the closed‑form OLS solution by default (not gradient descent).`

```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt 
```

### `Step 1: Load dataset`

```python
df = pd.read_csv("Study Hour.csv")
df.sample(5)
```

```python
X = df[['Hours']]   # X must be 2D for sklearn
y = df['Scores']
```

### `Step 2: Create and fit the model`

```python
model = LinearRegression()
model.fit(X, y)
```

### `Step 3: Get slope (m) and intercept (b)`

```python
m = model.coef_[0]
b = model.intercept_

print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
```

### `Step 4: Predict values`

```python
y_pred = model.predict(X)
df['Predicted_Y'] = y_pred
df.sample(5)
```

### `Step 5: Plot original points and best fit line`

```python
plt.scatter(X, y, color='blue', label="Actual Data")
plt.plot(X, y_pred, color='red', label='Best Fit Line')
plt.xlabel("Hours")
plt.ylabel("Score")
plt.legend()
plt.show()
```
