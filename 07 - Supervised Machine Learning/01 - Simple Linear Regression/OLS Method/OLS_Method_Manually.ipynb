## 🐍 OLS Method with CSV Dataset By Manually without using Built-in Function

```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
```

### `Step 1: Load dataset`

```python
df = pd.read_csv("Study Hour.csv")
df.sample(5)
```

```python
X = df["Hours"].values.astype(float)
y = df["Scores"].values.astype(float)
```

## `Step 2: Calculate means`

```python
x_mean = np.mean(X)
y_mean = np.mean(y)
print(f"Mean of x >> {x_mean} and Mean of y >> {y_mean}")
```

## `Step 3: Calculate covariance and variance`

```python
cov_xy = np.sum((X - x_mean) * (y - y_mean))
var_x = np.sum((X - x_mean) ** 2)
print(f"Covariance of x,y >> {cov_xy} and Variance of X >> {var_x}")
```

## `Step 4: Calculate slope (m) and intercept (b)`

```python
m = cov_xy / var_x
b = y_mean - m * x_mean

print(f"Slope (m): {m}")
print(f"Intercept (b): {b}")
```

## `Step 5: Predict values`

```python
y_pred = m * X + b
```

```python
print(y_pred)
```

## `Step 6: Add predictions to DataFrame and save if needed`

```python
df['Predicted_Y'] = y_pred
print(df.head())
```

## `Scatter plot of actual data`

```python
plt.scatter(df['Hours'], df['Scores'], color='blue', label='Actual Scores')
# Plot regression line
plt.plot(df['Hours'], df['Predicted_Y'], color='red', linewidth=2, label='Best Fit Line')

# Labels and legend
plt.xlabel("Hours Studied")
plt.ylabel("Scores")
plt.title("Hours vs Scores with Best Fit Line")
plt.legend()
plt.show()
```
